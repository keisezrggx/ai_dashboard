```python
import gspread
import pandas as pd
# from datetime import dates
from google.oauth2.service_account import Credentials
```

```python
SERVICE_ACCOUNT_FILE = '../../key/credentials.json'
SCOPES = ['https://www.googleapis.com/auth/spreadsheets',
          'https://www.googleapis.com/auth/drive']

creds = Credentials.from_service_account_file(SERVICE_ACCOUNT_FILE, scopes=SCOPES)
client = gspread.authorize(creds)

sheet = client.open("LEMBAR KERJA - DISLIKE").worksheet("Dislike Analyze (Old)")
data = sheet.get_all_values()

df = pd.DataFrame(data[1:], columns=data[0])
df.to_csv('../../raw_data/qc_kula.csv', index=False)
```

```python
df.info()
```

```python
# Copy columns by name so source column order changes do not break dashboard keys
required_columns = [
    'Score_date',
    'Business Type',
    'Code',
    'Main Category',
    'Checking Result (Sub Category)',
    'Background detail- ID',
    'Team/Category',
    'Chat ID',
]
missing_columns = [column for column in required_columns if column not in df.columns]
if missing_columns:
    raise KeyError(f"Missing required columns: {missing_columns}")

df_clean = df.loc[:, required_columns].copy()

# Data type
df_clean['Score_date'] = pd.to_datetime(
    df_clean['Score_date'], errors='coerce'
)

df_clean.info()
```

```python
df_clean.to_csv('../../dataset_kula/qc_kula.csv', index=False)
```
